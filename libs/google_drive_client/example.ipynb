# Google Drive Client example

The client maps paths relative to `local_root` into paths relative to `drive_root`.

```python
from google_drive_client import GoogleDriveClient, DriveClientInitArgs

client = GoogleDriveClient(
    DriveClientInitArgs(
        client_name='google_api',
        credentials_file='.credentials/credentials.json',
        root_folder_id='1LVUSQJjcuPEDt4taa_xgizsOih6vRTTb'
    )
)
```

```python
# List direct children
client.list_dir('/datasets')
```

```python
# List only files
client.list_files('/datasets')
```

```python
# Upload ./workspace/results/metrics.csv to /SAVIIA/results/metrics.csv
client.upload_to('/results/metrics.csv')
```
